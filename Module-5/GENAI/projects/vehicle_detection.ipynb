# Vehicle Object Detection

The provided `Images.zip` contains unlabeled photos and no bounding-box annotations. This notebook therefore demonstrates inference with a COCO-pretrained Faster R-CNN model; it does not train on these photos or report ground-truth detection accuracy. Add annotated boxes to enable supervised training and mAP evaluation.

```python
import sys
from pathlib import Path

ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / "Module-5" / "GENAI").is_dir()), None)
if ROOT is None:
    raise RuntimeError("Run this notebook from the repository or one of its subfolders.")
sys.path.insert(0, str(ROOT / "Module-5" / "GENAI" / "projects"))
from autonomous_driving import safe_extract_images, create_image_manifests
from autonomous_driving import load_vehicle_detector, detect_vehicles

archive = ROOT / "datasets" / "GEN AI" / "Capstone 1" / "Part 1" / "Images.zip"
artifacts = ROOT / "saved_models" / "genai" / "autonomous_driving"
images = safe_extract_images(archive, artifacts / "object_detection" / "images")
manifest = create_image_manifests(images, artifacts, seed=42)
print(f"Validated {len(images)} images; no annotations were present.")
manifest.groupby("split").size()
```

```python
import random
import torch
from IPython.display import display

torch, detector, transform, categories = load_vehicle_detector()
sample = random.Random(42).choice(images)
annotated, detections = detect_vehicles(sample, detector, transform, categories, torch, threshold=0.5)
display(annotated)
detections
```
