# Local AI Banner and Poster Generator

This notebook demonstrates the same image-generation function used by the Gradio application. Ollama locally refines the prompt and Diffusers generates an image on this PC; no OpenAI API key or ComfyUI server is used.

For the interactive application, run `image_generation.py` from the repository root.

```python
import os
import sys
from pathlib import Path

ROOT = next(
    (path for path in (Path.cwd(), *Path.cwd().parents)
     if (path / "Module-5" / "GENAI").is_dir()),
    None,
)
if ROOT is None:
    raise RuntimeError("Run this notebook from the repository or one of its subfolders.")
sys.path.insert(0, str(ROOT / "Module-5" / "GENAI" / "projects"))

from image_generation import generate_image

from lib.utility.genai.config import GenAIConfig
from lib.utility.genai.factory import get_llm

config = GenAIConfig.from_env()
llm = get_llm(config)
print(f"Ollama model is available: {config.chat_model}")
```

Edit the prompt to describe the subject, composition, palette, and style for the design. The Diffusers model is downloaded on first use and runs on CPU; the returned image is a Pillow RGB image.

```python
prompt = (
    "A polished summer music festival poster, vibrant sunset palette, "
    "silhouette of a seaside stage, clean editorial layout, no logos"
)
image, refined_prompt = generate_image(
    prompt, size="384x384", quality="Draft", llm=llm, return_prompt=True
)
print("Ollama-refined prompt:", refined_prompt)
display(image)
```
